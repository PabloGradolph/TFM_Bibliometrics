## **Código QR simple**
---
---

</span>

```python
# Enlace del codigo QR
url = "https://github.com/PabloGradolph/TFM_Bibliometrics"

# Crear el objeto QR
qr = qrcode.QRCode(
    version=1,
    error_correction=qrcode.constants.ERROR_CORRECT_H,
    box_size=10,
    border=4,
)

# Agregar el enlace
qr.add_data(url)
qr.make(fit=True)

# Crear la imagen con colores hexadecimales
img = qr.make_image(
    fill_color="#FFFFFF",
    back_color="#A6AAAC"
)

# Guardar la imagen
img.save("qr_simple.png")
```

## **Código QR con logo cuadrado**
---
---

</span>

```python
# Enlace del código QR
url = "https://github.com/PabloGradolph/TFM_Bibliometrics"

# Crear el objeto QR
qr = qrcode.QRCode(
    version=1,  # controla el tamaño del código (1 es el más pequeño)
    error_correction=qrcode.constants.ERROR_CORRECT_H,  # alto nivel de corrección de errores
    box_size=10,  # tamaño de cada caja en pixeles
    border=4,  # grosor del borde
)

# Agregar el enlace
qr.add_data(url)
qr.make(fit=True)

# Crear QR
qr_img = qr.make_image(fill_color="#000000", back_color="#FFFFFF").convert('RGB')

# Cargar logo (debe ser cuadrado y pequeño)
logo = Image.open("grafica.jpg")

# Calcular tamaño y posición del logo
qr_width, qr_height = qr_img.size
logo_size = qr_width // 5  # 1/4 del tamaño del QR
logo = logo.resize((logo_size, logo_size))

# Pegar logo en el centro del QR
pos = ((qr_width - logo_size) // 2, (qr_height - logo_size) // 2)
qr_img.paste(logo, pos)

# Guardar
qr_img.save("qr_square.png")
```

## **Código QR con logo circular**
---
---

</span>

```python
# Paso 1: Crear el código QR
url = "https://github.com/PabloGradolph/TFM_Bibliometrics"
qr = qrcode.QRCode(
  version=2,
  error_correction=qrcode.constants.ERROR_CORRECT_H,
  box_size=10,
  border=4,
)
qr.add_data(url)
qr.make(fit=True)

qr_img = qr.make_image(fill_color="#000000", back_color="#f8f2ec").convert('RGB')

# Paso 2: Cargar el logo y hacerlo circular
logo = Image.open("grafica.jpg").convert("RGBA")

# Redimensionar el logo
qr_width, qr_height = qr_img.size
logo_size = qr_width // 4
logo = logo.resize((logo_size, logo_size))

# Crear una máscara circular
mask = Image.new("L", (logo_size, logo_size), 0)
draw = ImageDraw.Draw(mask)
draw.ellipse((0, 0, logo_size, logo_size), fill=255)

# Crear una imagen nueva con fondo transparente
circular_logo = Image.new("RGBA", (logo_size, logo_size))
circular_logo.paste(logo, (0, 0), mask=mask)

# Paso 3: Pegar el logo en el centro del QR
pos = ((qr_width - logo_size) // 2, (qr_height - logo_size) // 2)
qr_img.paste(circular_logo, pos, mask=circular_logo)

# Paso 4: Guardar la imagen final
qr_img.save("qr_circular.png")
```
